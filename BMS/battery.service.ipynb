# Background Battery Management Service  

**Full code available at [battery.service](battery.service)**  
<br>  

This section documents the **Background Battery Management Service**, which ensures the Battery Management System (BMS) runs continuously in the background.  

The service is managed by **systemd**, allowing it to:  
- Automatically start the BMS script at boot.  
- Restart the script if it crashes or exits unexpectedly.  
- Log both debug messages and error messages to `journald` (or log files if configured by `setup.sh`).  
- Run independently from the user session to provide reliable background monitoring.  

> **NOTE:** The file paths in `battery.service` will be automatically overwritten by `setup.sh` to ensure the username matches the current machine’s username.  


## Let's Start Coding !
### The `[Unit]` section 
- In this section the Description of the service can be stated. 

### 2. The `[Service]` Section  

- **User**: Specifies the user account that will run the script in the background.  
- **Environment**: Defines the Battery Log paths so the Python script can access them.  
- **WorkingDirectory**: Sets the working directory (this can be modified as needed).  
- **ExecStart**: Defines the Python interpreter version and the path of the script to start.  
- **StandardError**: Redirects error logs to the system journal for later debugging.  
    ```ini
    StandardError=journal #Can be change to a specific log file 
    ```
- **StandardOutput**: Redirects debug/info logs to the system journal.
    ```ini
    StandardOutput=journal #Can be change to a specific log file 
    ```
- **Restart**: Ensures the service always restarts when the system starts or if the script crashes.
    ```ini
    Restart=always
    ```


### 3. The `[Install]` Section  

- **WantedBy=multi-user.target**  
  - Ensures the service is started automatically at boot.  
  - `multi-user.target` is the standard systemd target for non-graphical, multi-user systems (equivalent to traditional runlevel 3).  
  - By enabling the service with this target, the script will launch in the background whenever the system reaches this mode.  


### Save and Enable the Service  

The file will be saved as **`battery.service`**.  

- To make the service run in the background, the `.service` file must be placed under:  
  **`/etc/systemd/system/`**  

- After adding the file, reload the systemd daemon to apply the new service:  
    ```bash
    sudo systemctl daemon-reload
    ```

- Start the script using the systemctl command:
    ```bash
    sudo systemctl start battery.service
    ```
