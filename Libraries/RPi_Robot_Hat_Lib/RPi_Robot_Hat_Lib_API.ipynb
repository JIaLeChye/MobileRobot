# RPi Robot Hat Library - API Documentation
 
**Simple API Guide for Robot Control**

**Full Coding At [RPi_Robot_Hat_Lib.py](RPi_Robot_Hat_Lib.py)** 

This documentation covers only the **public API functions** that you need to control your robot. Internal functions are not documented here.

## Quick Start

```python
from RPi_Robot_Hat_Lib import RobotController

# Create robot instance
robot = RobotController(wheel_diameter=100, debug=False)

# Basic movement
robot.Forward(50)        # Move forward at 50% speed
robot.Backward(30)       # Move backward at 30% speed
robot.turn_left(40)      # Turn left at 40% speed
robot.stop()             # Stop all motors

# Cleanup when done
robot.cleanup()
```

## 1. Setup and Initialization

### RobotController(wheel_diameter=100, debug=False)
**Constructor to create a robot controller instance**

**Parameters:**
- `wheel_diameter` (int): Wheel diameter in millimeters (default: 100mm)
- `debug` (bool): Enable debug output (default: False)

**Example:**


```python
# Constructor example
robot = RobotController(wheel_diameter=120, debug=True)
```

### __version__()
**Get library version information**

**Returns:** Version string

**Example:**

```python
# Get library version
version = robot.__version__()
print(f"Library version: {version}")
```

## 2. Basic Movement API

### Forward(speed)
**Move robot forward**

**Parameters:**
- `speed` (int): Speed from 0-100 (percentage)

**Example:**


```python
# Move forward example
robot.Forward(50)  # Move forward at 50% speed
```

### Backward(speed)
**Move robot backward**

**Parameters:**
- `speed` (int): Speed from 0-100 (percentage)

**Example:**


```python
# Move backward example
robot.Backward(30)  # Move backward at 30% speed
```

### turn_left(speed)
**Turn robot left**

**Parameters:**
- `speed` (int): Turn speed from 0-100 (percentage)

**Example:**


```python
# Turn left example
robot.turn_left(40)  # Turn left at 40% speed
```

### turn_right(speed)
**Turn robot right**

**Parameters:**
- `speed` (int): Turn speed from 0-100 (percentage)

**Example:**

```python
# Turn right example
robot.turn_right(40)  # Turn right at 40% speed
```

### move(speed=0, turn=0)
**Advanced movement with speed and turn control**

**Parameters:**
- `speed` (int): Forward/backward speed from -100 to +100
- `turn` (int): Turn amount from -100 to +100 (negative=right, positive=left)

**Example:**

```python
# Advanced movement examples
robot.move(50, 20)   # Move forward at 50% speed, turning slightly left
robot.move(-30, 0)   # Move backward at 30% speed, straight
robot.move(0, -40)   # Turn right at 40% speed without moving forward
```

### Horizontal_Left(speed) & Horizontal_Right(speed)
**Mecanum wheel sideways movement (if your robot supports it)**

**Parameters:**
- `speed` (int): Speed from 0-100 (percentage)

**Example:**


```python
# Horizontal movement examples (mecanum wheels)
robot.Horizontal_Left(30)   # Move sideways left
robot.Horizontal_Right(30)  # Move sideways right
```

### stop() & Brake()
**Stop all motors immediately**

**Example:**


```python
# Stop motors examples
robot.stop()   # Stop all motors
robot.Brake()  # Same as stop()
```

## 3. Encoder Reading API

### get_encoder(motor, debug=False)
**Get current encoder count for a motor**

**Parameters:**
- `motor` (str): Motor name - 'RF', 'RB', 'LF', 'LB'
- `debug` (bool): Show debug information (optional)

**Returns:** Encoder count (integer)

**Example:**


```python
# Get encoder count example
count = robot.get_encoder('LF')  # Get left front encoder
print(f"Left front encoder: {count}")
```

### get_encoder_delta(motor, debug=False)
**Get encoder change since last read**

**Parameters:**
- `motor` (str): Motor name - 'RF', 'RB', 'LF', 'LB'
- `debug` (bool): Show debug information (optional)

**Returns:** Encoder delta (signed integer)

**Example:**


```python
# Get encoder delta example
delta = robot.get_encoder_delta('RF')  # Get change since last read
print(f"Right front moved: {delta} ticks")
```

### get_all_encoders()
**Get all encoder values at once**

**Returns:** Dictionary with all encoder values

**Example:**


```python
# Get all encoders example
encoders = robot.get_all_encoders()
print(f"All encoders: {encoders}")
# Output: {'RF': 1234, 'RB': 1200, 'LF': 1180, 'LB': 1190}
```

### get_rpm(motor, debug=False)
**Get motor RPM (revolutions per minute)**

**Parameters:**
- `motor` (str): Motor name - 'RF', 'RB', 'LF', 'LB'
- `debug` (bool): Show debug information (optional)

**Returns:** RPM value (float)

**Example:**


```python
# Get RPM example
rpm = robot.get_rpm('LF')
print(f"Left front RPM: {rpm:.2f}")
```

### get_distance(motor, debug=False)
**Get distance traveled by a motor in meters**

**Parameters:**
- `motor` (str): Motor name - 'RF', 'RB', 'LF', 'LB'
- `debug` (bool): Show debug information (optional)

**Returns:** Distance in meters (float)

**Example:**


```python
# Get distance example
distance = robot.get_distance('RF')
print(f"Right front traveled: {distance:.3f} meters")
```

### reset_encoders(debug=False)
**Reset all encoder counts to zero**

**Parameters:**
- `debug` (bool): Show reset confirmation (optional)

**Returns:** True if successful, False otherwise

**Example:**


```python
# Reset encoders example
success = robot.reset_encoders(debug=True)
if success:
    print("Encoders reset successfully")
```

## 4. Precise Movement API

### move_distance(distance, speed=40, debug=False)
**Move robot a specific distance**

**Parameters:**
- `distance` (float): Distance in meters (positive=forward, negative=backward)
- `speed` (int): Speed from 0-100 (default: 40)
- `debug` (bool): Show movement progress (optional)

**Returns:** Actual distance traveled (float)

**Example:**


```python
# Move specific distance examples
# Move forward 1 meter at 50% speed
actual = robot.move_distance(1.0, speed=50, debug=True)
print(f"Moved {actual:.3f} meters")

# Move backward 0.5 meters
robot.move_distance(-0.5, speed=30)
```

## 5. Servo Control API

### set_servo(servo_num, angle)
**Control servo motors**

**Parameters:**
- `servo_num` (int): Servo number (1 or 2)
- `angle` (int): Angle from 0-180 degrees

**Example:**


```python
# Servo control examples
robot.set_servo(1, 90)   # Set servo 1 to center position
robot.set_servo(2, 0)    # Set servo 2 to 0 degrees
robot.set_servo(1, 180)  # Set servo 1 to 180 degrees
```

## 6. Sensor Reading API

### read_line_sensors()
**Read digital line following sensors**

**Returns:** 5-bit value representing sensor states

**Example:**


```python
# Read line sensors example
sensors = robot.read_line_sensors()
print(f"Line sensors: {bin(sensors)}")

# Check individual sensors
if sensors & 0b10000:  # Leftmost sensor
    print("Left sensor detects line")
if sensors & 0b00001:  # Rightmost sensor
    print("Right sensor detects line")
```

### read_line_analog()
**Read analog line sensor value**

**Returns:** Analog value (0-255)

**Example:**


```python
# Read analog line sensor example
analog_value = robot.read_line_analog()
print(f"Analog line sensor: {analog_value}")
```

### get_battery()
**Read battery voltage**

**Returns:** Battery voltage in volts (float)

**Example:**

```python
# Read battery voltage example
voltage = robot.get_battery()
print(f"Battery voltage: {voltage:.1f}V")

if voltage < 6.5:
    print("Warning: Low battery!")
```

## 7. Sound/Buzzer API

### play_tone(frequency, duration)
**Play a tone on the buzzer**

**Parameters:**
- `frequency` (int): Frequency in Hz
- `duration` (float): Duration in seconds

**Example:**


```python
# Play tone examples
robot.play_tone(440, 1.0)    # Play A4 for 1 second
robot.play_tone(880, 0.5)    # Play A5 for 0.5 seconds
robot.play_tone(0, 0.2)      # Silence for 0.2 seconds
```

## 8. System Control API

### reset_system()
**Reset the entire robot hat system**

**Returns:** True if successful, False otherwise

**Example:**


```python
# Reset system example
success = robot.reset_system()
if success:
    print("System reset complete")
```

### cleanup()
**Clean up resources and stop all systems**

**Call this when you're done using the robot**

**Example:**


```python
# Cleanup example
robot.cleanup()  # Stop motors, center servos, cleanup GPIO
```

## 9. Complete Example Programs

### Basic Movement Test


```python
# Basic Movement Test Example
from RPi_Robot_Hat_Lib import RobotController
import time

robot = RobotController()

try:
    # Basic movement sequence
    robot.play_tone(440, 0.5)  # Start beep
    
    robot.Forward(50)           # Forward
    time.sleep(2)
    
    robot.turn_right(40)        # Turn right
    time.sleep(1)
    
    robot.Backward(30)          # Backward
    time.sleep(2)
    
    robot.stop()                # Stop
    
    robot.play_tone(880, 0.5)   # End beep
    
except KeyboardInterrupt:
    print("Stopped by user")
finally:
    robot.cleanup()
```

### Encoder Monitoring


```python
# Encoder Monitoring Example
from RPi_Robot_Hat_Lib import RobotController
import time

robot = RobotController(debug=True)

try:
    robot.reset_encoders()      # Reset to zero
    robot.Forward(40)           # Start moving
    
    for i in range(10):
        encoders = robot.get_all_encoders()
        print(f"Step {i+1}: {encoders}")
        time.sleep(0.5)
    
    robot.stop()
    
except KeyboardInterrupt:
    print("Stopped by user")
finally:
    robot.cleanup()
```

### Precise Distance Movement

```python
# Precise Distance Movement Example
from RPi_Robot_Hat_Lib import RobotController

robot = RobotController(wheel_diameter=100)

try:
    # Move exactly 1 meter forward
    distance_moved = robot.move_distance(1.0, speed=50, debug=True)
    print(f"Actually moved: {distance_moved:.3f} meters")
    
    # Move back to start position
    robot.move_distance(-distance_moved, speed=40)
    print("Returned to start position")
    
except KeyboardInterrupt:
    print("Stopped by user")
finally:
    robot.cleanup()
```

### Sensor and Battery Monitoring

```python
# Sensor and Battery Monitoring Example
from RPi_Robot_Hat_Lib import RobotController
import time

robot = RobotController()

try:
    while True:
        # Check battery
        voltage = robot.get_battery()
        print(f"Battery: {voltage:.1f}V", end=" | ")
        
        # Check line sensors
        line_digital = robot.read_line_sensors()
        line_analog = robot.read_line_analog()
        print(f"Line: {bin(line_digital)} (analog: {line_analog})")
        
        # Low battery warning
        if voltage < 6.5:
            robot.play_tone(220, 0.1)  # Low battery beep
        
        time.sleep(1)
        
except KeyboardInterrupt:
    print("\nStopped by user")
finally:
    robot.cleanup()
```

# API Summary

| Category           | Function(s)                                                                 | Description                                |
|--------------------|------------------------------------------------------------------------------|--------------------------------------------|
| **Movement**       | `Forward(speed)`, `Backward(speed)`                                          | Basic forward/backward movement            |
|                    | `turn_left(speed)`, `turn_right(speed)`                                      | Turning left or right                      |
|                    | `move(speed, turn)`                                                          | Combined movement + steering               |
|                    | `Horizontal_Left(speed)`, `Horizontal_Right(speed)`                          | Sideways (mecanum) movement                |
|                    | `stop()`, `Brake()`                                                          | Stop all motors                            |
|                    | `move_distance(distance, speed)`                                             | Move precise distance (meters)             |
| **Encoders**       | `get_encoder(motor)`                                                         | Get raw encoder count for a motor          |
|                    | `get_encoder_delta(motor)`                                                   | Change in encoder value since last read    |
|                    | `get_all_encoders()`                                                         | Get encoder counts for all motors          |
|                    | `get_rpm(motor)`                                                             | Calculate motor RPM                        |
|                    | `get_distance(motor)`                                                        | Distance traveled by a motor               |
|                    | `reset_encoders()`                                                           | Reset all encoder counts to zero           |
| **Servos & Sensors** | `set_servo(servo_num, angle)`                                               | Set servo position (0–180°)                |
|                    | `read_line_sensors()`                                                        | Read digital line sensors (5-bit)          |
|                    | `read_line_analog()`                                                         | Read analog line sensor value              |
|                    | `get_battery()`                                                              | Get battery voltage                        |
| **System Control** | `play_tone(frequency, duration)`                                             | Play tone on buzzer (Hz, seconds)          |
|                    | `reset_system()`                                                             | Reset coprocessor system                   |
|                    | `cleanup()`                                                                  | Clean shutdown of motors, servos, buzzer   |
|                    | `__version__()`                                                              | Print and return library version           |

---

⚡ **Reminder:** Always call `robot.cleanup()` when finished to properly stop motors, reset servos, and release resources.
