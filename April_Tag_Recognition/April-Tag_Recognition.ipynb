This is the documentation for April Tag Recognition using OpenCV and the apriltag library.

- time - For time-related operations

## Let's Start Coding ! 
### 1. Import all the required libraries 
- OpenCV (cv2) - For image processing and display
- Numpy (np) - For numerical operations
- Picamera2 - For Raspberry Pi camera control
- libcamera - For camera configuration
- apriltag - For AprilTag detection
- time - For time operations (optional, for FPS calculation)

````python
import cv2
import numpy as np
from picamera2 import Picamera2
from libcamera import controls
import apriltag
import time
````

### 2. Camera Configuration and Initialization

This setup configures and initializes the Raspberry Pi camera for optimal AprilTag detection.

- **Setup Name**: Camera Configuration and Initialization

- **Usage**:
  - Create Picamera2 object for camera interface:
    ```python
    cam = Picamera2()
    ```
  - Configure camera with specific parameters for AprilTag detection:
    ```python
    cam.configure(cam.create_preview_configuration(main={"format": 'XRGB8888', "size": (640, 480)}))
    ```
  - Start camera capture for real-time processing:
    ```python
    cam.start()
    ```
  - Enable continuous autofocus for better image quality:
    ```python
    cam.set_controls({"AfMode": controls.AfModeEnum.Continuous})
    ```

````python
cam = Picamera2()
cam.configure(cam.create_preview_configuration(main={"format": 'XRGB8888', "size": (640, 480)}))
cam.start()
cam.set_controls({"AfMode":controls.AfModeEnum.Continuous})
````

### 3. Main AprilTag Detection Loop

This function continuously captures frames, detects AprilTags, and displays detection results in real-time.

- **Function Name**: Main Detection Loop (while True)

- **Usage**:
  - **Capture current frame** from camera:
    ```python
    frame = cam.capture_array()
    ```
  - **Convert to grayscale** for AprilTag detection processing:
    ```python
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    ```
  - **Initialize AprilTag detector** object:
    ```python
    detector = apriltag.Detector()
    ```
  - **Detect AprilTags** in the grayscale image:
    ```python
    detections = detector.detect(gray)
    ```
  - **Process each detected AprilTag** and display information:
    ```python
    for detect in detections:
        print("tag_id: %s, center: %s" % (detect.tag_id, detect.center)) 
        cv2.putText(frame, "AprilTag Detected", (30, 480), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
    ```
  - **Display processed frame** with detection overlays:
    ```python
    cv2.imshow('frame', frame)
    ```
  - **Handle user exit** when 'q' key is pressed:
    ```python
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
    ```

````python
while True:
   
    frame = cam.capture_array()
    
    # Convert the captured frame to grayscale for AprilTag detection
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    
    # Create the AprilTag detector
    detector = apriltag.Detector()
    
    # Detect AprilTags in the grayscale image
    detections = detector.detect(gray)
    
    # Process each detected AprilTag
    for detect in detections:
        print("tag_id: %s, center: %s" % (detect.tag_id, detect.center)) 
        cv2.putText(frame, "AprilTag Detected", (30, 480), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
        
    # Display the frame with detection results
    cv2.imshow('frame', frame)

    # Exit when 'q' key is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
````

### 4. Program Cleanup and Resource Management

This workflow handles proper cleanup of camera and display resources when the program terminates.

- **Workflow Name**: Program Cleanup and Resource Management

- **Usage**:
  - **Stop camera operations** and release camera resources:
    ```python
    cam.stop()
    ```
  - **Close all OpenCV windows** and free GUI resources:
    ```python
    cv2.destroyAllWindows()
    ```
  - **Resource management benefits**:
    - Camera is properly released for other applications
    - All GUI windows are closed cleanly  
    - System resources are freed up properly

````python
cam.stop()
cv2.destroyAllWindows()
````
